# Abstract Factory — exercises

These build on the `Button`/`Checkbox`/`UIFactory` example. Re-run (or copy) the base classes below before starting, then complete each exercise in its own cell.

```python
from abc import ABC, abstractmethod


class Button(ABC):
    @abstractmethod
    def render(self) -> str:
        ...


class Checkbox(ABC):
    @abstractmethod
    def render(self) -> str:
        ...


class LightButton(Button):
    def render(self) -> str:
        return "[ Light Button ]"


class LightCheckbox(Checkbox):
    def render(self) -> str:
        return "[ ] Light Checkbox"


class DarkButton(Button):
    def render(self) -> str:
        return "[ Dark Button ]"


class DarkCheckbox(Checkbox):
    def render(self) -> str:
        return "[x] Dark Checkbox"


class UIFactory(ABC):
    @abstractmethod
    def create_button(self) -> Button:
        ...

    @abstractmethod
    def create_checkbox(self) -> Checkbox:
        ...


class LightThemeFactory(UIFactory):
    def create_button(self) -> Button:
        return LightButton()

    def create_checkbox(self) -> Checkbox:
        return LightCheckbox()


class DarkThemeFactory(UIFactory):
    def create_button(self) -> Button:
        return DarkButton()

    def create_checkbox(self) -> Checkbox:
        return DarkCheckbox()


def render_ui(factory: UIFactory) -> None:
    print(factory.create_button().render())
    print(factory.create_checkbox().render())
```

## Exercise 2 — Add a new product kind (adaptation)

**Goal:** Add a `Slider` product kind to the family. Every existing concrete factory (`LightThemeFactory`, `DarkThemeFactory`, and your `HighContrastThemeFactory` from Exercise 1) must be updated to produce a matching `Slider`, and `render_ui` must render it too.

**Acceptance criteria:**
- `UIFactory` gains a `create_slider()` abstract method.
- Each concrete factory implements it and returns a `Slider` implementation consistent with its family (e.g., `DarkSlider` for `DarkThemeFactory`).
- `render_ui` prints the slider alongside the button and checkbox.
- Confirm all three factories still run without error after the change.

```python
# Your solution here
from abc import ABC, abstractmethod


class Button(ABC):
    @abstractmethod
    def render(self) -> str:
        ...


class Checkbox(ABC):
    @abstractmethod
    def render(self) -> str:
        ...

class Slider(ABC):
    @abstractmethod 
    def render(self) -> str:
        ...

class LightButton(Button):
    def render(self) -> str:
        return "[ Light Button ]"

class LightCheckbox(Checkbox):
    def render(self) -> str:
        return "[ ] Light Checkbox"

class LightSlider(Slider):
    def render(self) -> str:
        return "[------] Light Slider"

#------------------------

class DarkButton(Button):
    def render(self) -> str:
        return "[ Dark Button ]"


class DarkCheckbox(Checkbox):
    def render(self) -> str:
        return "[ ] Dark Checkbox"

class DarkSlider(Slider):
    def render(self) -> str:
        return "[------] Dark Slider"

#New button and checkbox 
class HighContrastButton(Button):
    def render(self) -> str:
        return "[ High Contrast Button ]"

class HighContrastCheckbox(Checkbox):
    def render(self) -> str:
        return "[x] High Contrast Checkbox"

class HighContrastSlider(Slider):
    def render(self) -> str:
        return "[------] High Contrast Slider"


class UIFactory(ABC):
    @abstractmethod
    def create_button(self) -> Button:
        ...

    @abstractmethod
    def create_checkbox(self) -> Checkbox:
        ...

    @abstractmethod
    def create_slider(self) -> Slider:
        ... 


class LightThemeFactory(UIFactory):
    def create_button(self) -> Button:
        return LightButton()

    def create_checkbox(self) -> Checkbox:
        return LightCheckbox()

    def create_slider(self) -> Slider:
        return LightSlider()


class DarkThemeFactory(UIFactory):
    def create_button(self) -> Button:
        return DarkButton()

    def create_checkbox(self) -> Checkbox:
        return DarkCheckbox()

    def create_slider(self) -> Slider:
        return DarkSlider()


#New theme factory 
class HighContrastThemeFactory(UIFactory):
    def create_button(self) -> Button:
        return HighContrastButton()

    def create_checkbox(self) -> Checkbox:
        return HighContrastCheckbox()

    def create_slider(self) -> Slider:
        return HighContrastSlider()

def render_ui(factory: UIFactory) -> None:
    print(factory.create_button().render())
    print(factory.create_checkbox().render())
    print(factory.create_slider().render())
```

```python
def render_ui(factory: UIFactory) -> None:
    button = factory.create_button()
    checkbox = factory.create_checkbox()
    print(button.render())
    print(checkbox.render())
    slider = factory.create_slider()
    print(slider.render())


print("Light theme:")
render_ui(LightThemeFactory())

print("\nDark theme:")
render_ui(DarkThemeFactory())

print("\nHigh Contrast theme:")
render_ui(HighContrastThemeFactory())
```
