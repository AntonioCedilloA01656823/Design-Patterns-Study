# Abstract Factory — UI theme example

We build two product kinds, `Button` and `Checkbox`, each with a `Light` and `Dark` family variant.
A `UIFactory` abstract factory produces one matched pair per family, so the client never mixes a light button with a dark checkbox.

```python
from abc import ABC, abstractmethod


# --- Abstract products ---

class Button(ABC):
    @abstractmethod
    def render(self) -> str:
        ...


class Checkbox(ABC):
    @abstractmethod
    def render(self) -> str:
        ...
```

```python
# --- Concrete products: Light family ---

class LightButton(Button):
    def render(self) -> str:
        return "[ Light Button ]"


class LightCheckbox(Checkbox):
    def render(self) -> str:
        return "[ ] Light Checkbox"


# --- Concrete products: Dark family ---

class DarkButton(Button):
    def render(self) -> str:
        return "[ Dark Button ]"


class DarkCheckbox(Checkbox):
    def render(self) -> str:
        return "[x] Dark Checkbox"
```

```python
# --- Abstract factory ---

class UIFactory(ABC):
    @abstractmethod
    def create_button(self) -> Button:
        ...

    @abstractmethod
    def create_checkbox(self) -> Checkbox:
        ...


# --- Concrete factories, one per family ---

class LightThemeFactory(UIFactory):
    def create_button(self) -> Button:
        return LightButton()

    def create_checkbox(self) -> Checkbox:
        return LightCheckbox()


class DarkThemeFactory(UIFactory):
    def create_button(self) -> Button:
        return DarkButton()

    def create_checkbox(self) -> Checkbox:
        return DarkCheckbox()
```

```python
# --- Client: depends only on UIFactory / Button / Checkbox, never on concrete classes ---

def render_ui(factory: UIFactory) -> None:
    button = factory.create_button()
    checkbox = factory.create_checkbox()
    print(button.render())
    print(checkbox.render())


print("Light theme:")
render_ui(LightThemeFactory())

print("\nDark theme:")
render_ui(DarkThemeFactory())
```

Notice `render_ui` never mentions `LightButton`, `DarkCheckbox`, etc. Swapping the whole visual family is a one-line change: pass a different factory instance. This is the defining behavior of Abstract Factory — the client is decoupled from concrete classes and family consistency is guaranteed by construction.
