## Exercise 3 (critique): When composition adds no value

Below, `Wallet` composes a `Balance` object but forwards every single method to it with no added behavior.

**Goal:** Explain in a markdown cell whether this composition is justified, and propose a fix (either add real value to `Wallet`, or eliminate the indirection).

**Acceptance criteria:**
- You identify this as the "thin wrapper" anti-pattern from the composition summary.
- You propose a concrete fix: either give `Wallet` genuine added behavior (e.g., validation, logging, currency conversion) or remove the wrapper and use `Balance` directly.
- You state one scenario where keeping a thin wrapper like this would actually be justified (hint: future extensibility, or hiding a dependency you plan to replace).

```python
class Balance:
    def __init__(self, amount: float):
        self.amount = amount

    def deposit(self, value: float) -> None:
        self.amount += value

    def withdraw(self, value: float) -> None:
        self.amount -= value


class Wallet:
    def __init__(self, balance: Balance):
        self._balance = balance

    def deposit(self, value: float) -> None:
        self._balance.deposit(value)

    def withdraw(self, value: float) -> None:
        self._balance.withdraw(value)
```
