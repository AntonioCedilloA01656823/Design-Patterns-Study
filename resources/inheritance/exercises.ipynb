## Exercise 3 (critique): Spot the Liskov Substitution Principle violation

Below, `Rectangle` has settable `width`/`height`, and `Square` inherits from it but forces both to stay equal by overriding the setters.

**Goal:** Explain in a markdown cell why `Square` here is *not* safely substitutable for `Rectangle`, using a concrete example of code that breaks. Then decide: would composition avoid this problem, and how?

**Acceptance criteria:**
- Your example shows a piece of client code that behaves correctly for any `Rectangle` but incorrectly for a `Square` passed in its place.
- Your answer references the Liskov Substitution Principle by name.
- You state whether composition would sidestep the issue and why (or why not).

```python
class MutableRectangle:
    def __init__(self, width: float, height: float):
        self.width = width
        self.height = height

    def area(self) -> float:
        return self.width * self.height


class MutableSquare(MutableRectangle):
    def __init__(self, side: float):
        super().__init__(side, side)

    def __setattr__(self, key, value):
        # Keep width and height equal whenever either is set
        super().__setattr__(key, value)
        if key in ("width", "height"):
            super().__setattr__("width", value)
            super().__setattr__("height", value)


def double_the_width(rect: MutableRectangle) -> None:
    original_height = rect.height
    rect.width *= 2
    assert rect.height == original_height, "height should be unaffected by width change!"


# Try calling double_the_width with a MutableRectangle, then with a MutableSquare.
```
