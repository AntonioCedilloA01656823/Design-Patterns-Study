# Inheritance — Example

We model shapes with a `Shape` base class that defines a **template method** `describe()`. Subclasses only override `area()`; `describe()` is inherited unchanged, demonstrating overriding, `super()`, and polymorphism.

```python
class Shape:
    def __init__(self, name: str):
        self.name = name

    def area(self) -> float:
        raise NotImplementedError

    def describe(self) -> str:
        # Template method: stable logic that relies on the overridden area()
        return f"{self.name} has an area of {self.area():.2f}"
```

```python
import math


class Circle(Shape):
    def __init__(self, radius: float):
        super().__init__("Circle")
        self.radius = radius

    def area(self) -> float:
        return math.pi * self.radius ** 2
```

```python
class Rectangle(Shape):
    def __init__(self, width: float, height: float):
        super().__init__("Rectangle")
        self.width = width
        self.height = height

    def area(self) -> float:
        return self.width * self.height
```

```python
shapes: list[Shape] = [Circle(3), Rectangle(4, 5)]

for shape in shapes:
    # Client code only knows about Shape; describe() is inherited, area() is overridden
    print(shape.describe())
```

## Extending behavior with `super()`

A `Square` is a specialized `Rectangle`. Instead of duplicating `Rectangle`'s logic, it calls `super().__init__()` and only adds the constraint that width equals height.

```python
class Square(Rectangle):
    def __init__(self, side: float):
        super().__init__(width=side, height=side)
        self.name = "Square"


print(Square(4).describe())
```

## Reflection

- `describe()` is written once in `Shape` and reused unchanged by every subclass — no duplication.
- Each subclass only overrides the one part that varies (`area()`), following the template method pattern.
- `Square` reuses `Rectangle`'s implementation via `super()` instead of copy-pasting it, showing inheritance can chain sensibly when the "is-a" relationship holds at every level (`Square` is a `Rectangle` is a `Shape`).
- This works well because every subclass satisfies the same contract (`area() -> float`) and can be substituted for `Shape` anywhere it's used (Liskov Substitution Principle).
